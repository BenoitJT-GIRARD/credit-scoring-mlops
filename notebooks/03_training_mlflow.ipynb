# 03 — Comparing models, with MLflow tracking

## What this notebook is for

Comparing several credit-scoring models under cross-validation, and picking the one that goes on to tuning.

## How the runs were produced

By scripts, not by this notebook — so that the comparison can be re-run:

- `uv run python scripts/train_mlflow.py --cv 5`
- `uv run python scripts/train_mlflow.py --cv 5 --undersample`

Four metrics are tracked per run:

- `roc_auc_mean`
- `pr_auc_mean`
- `business_cost_per_row`
- `best_threshold_mean`

**The selection criterion is `business_cost_per_row`**, not AUC. On an imbalanced problem where a missed default and a refused good customer cost different amounts, ranking quality alone does not decide anything — the cost does, and the threshold that minimises it is part of the model.

To browse the runs:

````python
from pathlib import Path
import mlflow
from mlflow.tracking import MlflowClient

# ---- Find the project root by looking for the 'src' folder ----
cwd = Path().resolve()
root = cwd

while root != root.parent and not (root / "src").exists():
    root = root.parent

if not (root / "src").exists():
    raise RuntimeError("Impossible de trouver la racine du projet (dossier 'src' introuvable).")

# ---- The real SQLite store ----
db_path = (root / "mlflow" / "mlflow.db").resolve()

if not db_path.exists():
    raise RuntimeError(f"Base MLflow introuvable : {db_path}")

TRACKING_URI = f"sqlite:///{db_path.as_posix()}"

mlflow.set_tracking_uri(TRACKING_URI)
client = MlflowClient()

print("Tracking URI:", mlflow.get_tracking_uri())
print("Experiments disponibles:", [e.name for e in client.search_experiments()])
````

````python
EXP_NAME = "credexp"

exp = mlflow.get_experiment_by_name(EXP_NAME)

if exp is None:
    raise RuntimeError(
        f"Experiment '{EXP_NAME}' introuvable. "
        f"Disponibles: {[e.name for e in client.search_experiments()]}"
    )

print("Exp utilisée :", exp.name, "| id:", exp.experiment_id)
````

## Which model goes to tuning

From the cross-validation:

- The dummy baselines are beaten by a wide margin, as they should be.
- Logistic regression holds up as a solid baseline.
- LightGBM gives the best cost / stability trade-off.
- The MLP trails on this tabular data, which is the usual outcome and not a surprise worth pursuing.

**Retained before tuning: LightGBM with undersampling.**

On three grounds: the lowest business cost per row, competitive ROC-AUC and PR-AUC, and — the one that is easy to skip — stability across folds. A model that wins on the mean and swings across folds is a model whose next fold is unknown.
