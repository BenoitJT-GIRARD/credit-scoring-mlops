# 04 — Optuna tuning, the final model, and the MLflow Model Registry

## What this notebook is for
1. Show the Optuna fine-tuning of LightGBM.
2. Compare the final model before and after tuning, under the same protocol on the same holdout.
3. Confirm the model is versioned in the Model Registry (v1 → v2).

## One trap with the business-cost metric

The raw `business_cost` scales with the number of rows evaluated, so it is only comparable between runs that scored the same number of them. In practice:

- between models in notebook 03 — same cross-validation, so comparable;
- between Optuna trials — same folds, same size, same objective, so comparable;
- between final models before and after tuning — only because the holdout is identical.

Comparing a cross-validated cost against a holdout cost, or two costs computed on different sample sizes, produces a number that looks like a comparison and is not one.

````python
from pathlib import Path
import mlflow
from mlflow.tracking import MlflowClient
import pandas as pd

# Find the project root by looking for the 'src' folder
cwd = Path().resolve()
root = cwd
while root != root.parent and not (root / "src").exists():
    root = root.parent
if not (root / "src").exists():
    raise RuntimeError("Impossible de trouver la racine du projet (dossier 'src' introuvable).")

db_path = (root / "mlflow" / "mlflow.db").resolve()
if not db_path.exists():
    raise RuntimeError(f"DB MLflow introuvable : {db_path}")

TRACKING_URI = f"sqlite:///{db_path.as_posix()}"
mlflow.set_tracking_uri(TRACKING_URI)
client = MlflowClient()

print("Tracking URI:", mlflow.get_tracking_uri())
print("Experiments:", [e.name for e in client.search_experiments()])
````

````python
EXP_NAME = "credexp"
exp = mlflow.get_experiment_by_name(EXP_NAME)

if exp is None:
    exps = client.search_experiments()
    non_default = [e for e in exps if e.name != "Default"]
    if not non_default:
        raise RuntimeError("Aucune expérience non-Default trouvée.")
    exp = non_default[0]
    print("EXP_NAME introuvable, fallback ->", exp.name)

print("Using experiment:", exp.name, "| id:", exp.experiment_id)
````

````python
runs = mlflow.search_runs([exp.experiment_id], output_format="pandas")
print("Total runs:", len(runs))
display(runs[["run_id", "start_time", "status", "tags.mlflow.runName"]].head(20))
````

## 1) Optuna fine-tuning of LightGBM

Run by script, so it can be repeated:
- `uv run python scripts/tune_optuna.py --trials 30 --cv 3`

What is read back from the tuning runs:
- the best business cost reached
- the hyperparameters that produced it
- how much the trials disagreed, when that was logged — a best trial that sits far from the rest is a best trial that may not repeat

````python
# Heuristic: the run name contains 'optuna' or 'tune'
mask_tune = runs["tags.mlflow.runName"].fillna("").str.lower().str.contains("optuna|tune")
tune_runs = runs[mask_tune].copy().sort_values("start_time", ascending=False)

print("Tuning runs found:", len(tune_runs))
display(tune_runs[["run_id", "start_time", "tags.mlflow.runName"]].head(20))
````

````python
# The columns that matter, when they are present
cols = [
    "run_id",
    "start_time",
    "tags.mlflow.runName",
    "metrics.best_value",
    "metrics.best_business_cost",
    "metrics.business_cost_per_row",
]
cols = [c for c in cols if c in tune_runs.columns]
display(tune_runs[cols].head(10))
````

````python
# Most recent by default
if len(tune_runs) == 0:
    print("Aucun run Optuna détecté. (OK si tu n'as pas loggé Optuna dans MLflow.)")
else:
    run_id = tune_runs.iloc[0]["run_id"]
    data = runs[runs["run_id"] == run_id].iloc[0]

    metric_cols = sorted([c for c in runs.columns if c.startswith("metrics.")])
    param_cols = sorted([c for c in runs.columns if c.startswith("params.")])

    print("Selected tune run:", run_id)
    display(pd.DataFrame({"metric": metric_cols, "value": [data.get(c) for c in metric_cols]}).dropna())
    display(pd.DataFrame({"param": param_cols, "value": [data.get(c) for c in param_cols]}).dropna())
````

## 2) The final models, before and after tuning

A final model is trained on a development set and scored on a holdout it has never seen. Only final runs are compared with each other, so that the protocol is held fixed and the tuning is the only thing that differs.

```
uv run python scripts/train_final.py --run-name lgbm_final_v1
uv run python scripts/train_final.py --run-name lgbm_final_v2
```

The candidate carried forward from notebook 03 is LightGBM with undersampling; what is measured here is what the tuning added to it.

````python
mask_final = runs["tags.mlflow.runName"].fillna("").str.lower().str.contains("final")
final_runs = runs[mask_final].copy().sort_values("start_time", ascending=False)

print("Final runs found:", len(final_runs))
display(final_runs[["run_id", "start_time", "tags.mlflow.runName"]].head(20))
````

````python
cols_final = [
    "run_id",
    "start_time",
    "tags.mlflow.runName",
    "metrics.val_business_cost",
    "metrics.val_best_threshold",
    "metrics.val_roc_auc",
    "metrics.val_pr_auc",
    "metrics.holdout_business_cost",
    "metrics.holdout_roc_auc",
    "metrics.holdout_pr_auc",
    "metrics.holdout_tn",
    "metrics.holdout_fp",
    "metrics.holdout_fn",
    "metrics.holdout_tp",
]
cols_final = [c for c in cols_final if c in final_runs.columns]

df_final = final_runs[cols_final].copy()

# Sort by holdout cost when it is available, otherwise by date
if "metrics.holdout_business_cost" in df_final.columns:
    df_final = df_final.sort_values("metrics.holdout_business_cost", ascending=True)

display(df_final.head(10))
````

````python
if len(df_final) == 0:
    raise RuntimeError("Aucun run final trouvé. Lance ton train_final puis re-run ce notebook.")

# The two rows kept are the two best by holdout cost, or the two most recent
# when that column is missing. Either way they are the baseline and the tuned run.
candidates = df_final.head(2).copy()
display(candidates)

print("Selected finals:")
for _, r in candidates.iterrows():
    print("-", r["tags.mlflow.runName"], "| run_id:", r["run_id"])
````

### Which final model is kept

The one with the lowest business cost on the holdout, at identical protocol, provided its ROC-AUC and PR-AUC have not collapsed to buy that cost — a threshold can always be moved to make one number look good at the expense of the others.

Retained: **LightGBM with undersampling, tuned.**

## 3) The Model Registry, as evidence of versioning

Checking that the final model really is registered: the same model name, and two distinct versions — v1 for the baseline, v2 for the tuned one. A registry entry is what lets the serving side ask for a version rather than a file path.

````python
import mlflow
from mlflow.tracking import MlflowClient

# Use the same tracking URI verified above
mlflow.set_tracking_uri(TRACKING_URI)
mlflow.set_registry_uri(TRACKING_URI)  # registry = même DB sqlite

client = MlflowClient()

models = client.search_registered_models()
print("Registered models:", [m.name for m in models])

# Set the exact registered name here if it is known;
# otherwise list what the registry holds first.
````

````python
MODEL_NAME = "credit_scoring_model"

versions = client.search_model_versions(f"name='{MODEL_NAME}'")
if len(versions) == 0:
    raise RuntimeError(
        f"Aucune version trouvée pour '{MODEL_NAME}'. "
        "Vérifie que train_final loggue bien registered_model_name."
    )

# Sort numerically by version
versions_sorted = sorted(versions, key=lambda v: int(v.version))

for v in versions_sorted:
    print(f"v{v.version} | run_id={v.run_id} | status={v.status} | stage={getattr(v, 'current_stage', '')}")
````

## Where this leaves the model

- Notebook 03 chose the model family under cross-validation: LightGBM with undersampling.
- Notebook 04 tuned it and measured what the tuning bought on a fixed holdout.
- The retained model is registered in MLflow, which is what the next two steps consume: explainability (gain importance and SHAP), and serving.
